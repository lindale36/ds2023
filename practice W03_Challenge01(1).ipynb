### Task 4

Use the `.plot()` method on the dataframe to create a line graph showing `date` on the x axis, the stocks' value on the y axis, and a separate plot for each stock symbol.

- Make the figures $10$ inches wide by $5$ inches tall.

````python
stocks1 = stocks.set_index('date').plot(rot=0, figsize=(10,5), subplots=True, sharey=True)
plt.show()
````

### Task 5

Plot the mean stock price over time.

Give your plot a title.

````python
stocks1.mean(axis=1).plot(figsize=(10,5), title="Mean Stock Price Over Time")
plt.show()
````
