## The data set

Today's dataset is `tips`, available from Seaborn.
It records $244$ observations of restaurant bills and tips collected by a waiter at an American restaurant over several months.

Each row is one dining party.

**Data dictionary**

| Variable | Data Type | Domain / Range | Description |
|---|---|---|---|
| total_bill | Continuous | $3.07 – $50.81 | Total bill in US dollars. |
| tip | Continuous | $1.00 – $10.00 | Tip amount in US dollars. |
| sex | Categorical | Female, Male | Sex of the person who paid. |
| smoker | Categorical | Yes, No | Whether the party included a smoker. |
| day | Categorical | Thur, Fri, Sat, Sun | Day of the week. |
| time | Categorical | Lunch, Dinner | Meal period. |
| size | Integer | 1 – 6 | Size of the dining party. |

We will work mostly with `total_bill` and `tip`, both of which are monetary amounts and therefore good candidates for a Q-Q plot against the normal distribution.
