## Worked Example

**Starting question:** How do price and expert rating relate to each other in this dataset?

### Step 1: Task Level

We first turn the starting question into a task-level question.

> **Task:** How does a wine's retail price relate to the rating it receives from expert reviewers?

A wine buyer might ask this about the market.

Note that this question names no chart type and no interaction. It could be answered with a scatter plot, a heatmap, a box plot, or a line chart — which confirms it belongs at the task level.

### Step 2: Sub-Task Level

We break the task into two concrete analytical questions, each approaching it from a different angle.

| | Sub-task question |
|---|---|
| Sub-task 1 | How are wine prices distributed across the dataset? |
| Sub-task 2 | Do expert ratings tend to rise as prices increase? |

Both sub-tasks address the same task from different angles.

- Sub-task 1 examines the shape of one variable
- Sub-task 2 examines how two variables relate

```{note}
In the EDA Demo, the first sub-task led to the discovery that **prices cluster at five-dollar intervals** —
a pattern invisible in a plain histogram but visible in a heatmap and bar chart.

Defining the sub-task precisely first would have guided you toward the right chart sooner.
```

### Step 3: Operation Level

We translate each sub-task into an Action–Target pair, using an Amar et al. primitive as the Action.

| Sub-task | Operation question | Action (Amar) | Target |
|----------|--------------------|---------------|--------|
| How are wine prices distributed? | What is the full range of wine prices in the dataset? | Determine Range | Price |
| Do ratings rise with price? | Do wine prices and expert ratings move together? | Correlate | Price, Points |

The two operations use different Amar primitives.

- Determine Range establishes the extent of prices; 

- Correlate tests whether two variables move together.
