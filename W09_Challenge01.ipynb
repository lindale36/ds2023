### Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ipywidgets import interact
import ipywidgets as widgets

sns.set_style('white')
```

### Load Data

```python
data_url = "https://virginia.box.com/shared/static/eqpwseity23af9la6ppowkgv2ho4rvmz.csv"
reviews = pd.read_csv(data_url).set_index("obs_id")

# Filter to the 7 most-reviewed countries and drop rows with missing price
top_countries = reviews.value_counts('country').index[:7]
reviews = reviews.loc[reviews['country'].isin(top_countries)].dropna(subset=['price'])
reviews[['country', 'taster']] = reviews[['country', 'taster']].astype('category')

review_groups = ['country', 'variety', 'taster']
review_measures = ['price', 'points']

reviews.head()
```

Begin responses here.

```python
@interact(ascending=False)
def plot_countries(ascending=False):
    reviews['country'].value_counts().sort_values(ascending=ascending).plot.barh()
    plt.xlabel('Number of Reviews')
    plt.show()
```

```python
@interact(col=review_groups, ascending=False)
def plot_group(col='country',ascending=False):
    reviews[col].value_counts().head(10).sort_values(ascending=ascending).plot.barh()
    plt.title(col)
    plt.show()
```
