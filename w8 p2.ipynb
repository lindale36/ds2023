```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
from seaborn import load_dataset

results = pd.read_csv("chatgpt_detection_results.csv")

results.info()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
for key, group in results.groupby('actual_type'):
    group['detection_probability'].plot.hist(ax=ax, label=key, alpha=0.5, bins=20)

ax.set_title("Distribution of Detection Probability")
ax.set_xlabel("Detection Probability")
ax.set_ylabel("Frequency")
ax.legend()
plt.tight_layout()
plt.show()
```
